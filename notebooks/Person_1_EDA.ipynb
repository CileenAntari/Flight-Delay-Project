## 3. Import Libraries

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: f'{x:.2f}')

print('[OK] Libraries imported successfully!')
```

## 4. Load Dataset

```
df = pd.read_csv('../data/raw/flight_delays.csv')
print(f'[OK] Data loaded: {df.shape[0]:,} rows, {df.shape[1]:,} columns')
```

## 6. Dataset Dimensions

## 7. Column Names

```
print()
print('='*70)
print('COLUMN NAMES')
print('='*70)
for idx, col in enumerate(df.columns, 1):
    print(f'{idx:2d}. {col}')
```

## 8. Data Type Distribution

```
print()
print('='*70)
print('DATA TYPE DISTRIBUTION')
print('='*70)
for dtype, count in df.dtypes.value_counts().items():
    print(f'{str(dtype):20s}: {count} columns')
```

## 9. Dataset Description Table

```
desc = pd.DataFrame({'Column': df.columns, 'Type': df.dtypes.values, 'Missing': df.isnull().sum().values})
print(desc.to_string(index=False))
```

## 10. Missing Values Analysis

```
completeness = (1 - (df.isnull().sum().sum() / (len(df) * len(df.columns)))) * 100
print(f'Data Completeness: {completeness:.2f}%')
```

## 11. Duplicate Records

```
dups = df.duplicated().sum()
print(f'Duplicates: {dups}')
if dups == 0:
    print('[OK] No duplicates found')
```

## 12. Descriptive Statistics

```
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols].describe()
```

## 13. Delay Analysis

```
delay_stats = df['DelayMinutes'].describe()
delayed = (df['DelayMinutes'] > 0).sum()
print(f'Mean Delay: {delay_stats["mean"]:.2f} minutes')
print(f'Delayed Flights: {delayed:,}')
```

## 14. Outlier Detection

```
Q1 = df['DelayMinutes'].quantile(0.25)
Q3 = df['DelayMinutes'].quantile(0.75)
IQR = Q3 - Q1
outliers = ((df['DelayMinutes'] < Q1 - 1.5*IQR) | (df['DelayMinutes'] > Q3 + 1.5*IQR)).sum()
print(f'Outliers: {outliers:,}')
```

## 15. Airport Analysis

```
origins = df['Origin'].unique()
destinations = df['Destination'].unique()
all_airports = sorted(set(list(origins) + list(destinations)))
print(f'Total Airports: {len(all_airports)}')
print(f'Codes: {", ".join(all_airports)}')
```

## 16. Airline Distribution

```
airline_counts = df['Airline'].value_counts()
for airline, count in airline_counts.items():
    pct = (count / len(df)) * 100
    print(f'{airline}: {count:,} ({pct:.2f}%)')
```
